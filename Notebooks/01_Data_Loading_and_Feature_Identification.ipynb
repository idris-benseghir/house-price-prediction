##### **Section 1 — Import Libraries**

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

##### **Section 2 — Load Dataset**

```python
train_ds = r"C:\Users\idris\OneDrive\Desktop\Master_AI\Applied_ML\Assignment 1\house-prices-advanced-regression-techniques\train.csv"

data = pd.read_csv(train_ds)

display(data.head())

print("Dataset shape:", data.shape)
```

##### **Section 3 — Identify Numerical and Categorical Features**

```python
numerical_features = data.select_dtypes(include=np.number).columns.tolist()

categorical_features = data.select_dtypes(include=['object', 'category']).columns.tolist()

print("Number of numerical features:", len(numerical_features))
print("Number of categorical features:", len(categorical_features))

print("\nNumerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

##### **Section 4 — Save Feature Names into Text Files**

```python
with open(r"C:\Users\idris\OneDrive\Desktop\Master_AI\Applied_ML\Assignment 1\house-prices-advanced-regression-techniques\numerical_features.txt", "w") as file:
    file.write("\n".join(numerical_features))

with open(r"C:\Users\idris\OneDrive\Desktop\Master_AI\Applied_ML\Assignment 1\house-prices-advanced-regression-techniques\categorical_features.txt", "w") as file:
    file.write("\n".join(categorical_features))

print("Feature names saved successfully!")
```
