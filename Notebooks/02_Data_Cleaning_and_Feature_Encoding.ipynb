#### **Main Operations**
- Load the original training dataset and previously identified feature lists.
- Remove duplicate records.
- Split the dataset into training (80%) and testing (20%) sets.
- Handle missing values in structural categorical features by replacing them with `"None"` where appropriate.
- Remove rows containing missing values in categorical features.
- Apply ordinal encoding to ordered categorical features and binary encoding to binary features.
- Apply One-Hot Encoding to nominal categorical features.
- Handle missing numerical values:
  - Remove rows with missing `GarageYrBlt`.
  - Apply median imputation to `LotFrontage` and `MasVnrArea`, using training-set medians.
- Align training and testing feature columns to ensure consistency.
- Verify remaining missing values and feature consistency.
- Save the cleaned training and testing datasets.

#### **Output Files**
- `clean_train.csv`
- `clean_test.csv`

These datasets will be used in subsequent notebooks for exploratory analysis, feature selection, numerical scaling, and regression model development.

**Note:** The train/test split is performed before data analysis and preprocessing to reduce the risk of data leakage and ensure a more reliable evaluation of model performance.

##### **Section 1 — Import Libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

##### **Section 2 — Load Dataset**

```python
train_ds = r"C:\Users\idris\OneDrive\Desktop\Master_AI\Applied_ML\Assignment 1\house-prices-advanced-regression-techniques\train.csv"

train_df = pd.read_csv(train_ds)

print("Dataset Shape:", train_df.shape)

display(train_df.head())

train_df.info()
```

##### **Section 3 — Load Feature Names**

```python
folder = r"C:\Users\idris\OneDrive\Desktop\Master_AI\Applied_ML\Assignment 1\house-prices-advanced-regression-techniques"

# Read numerical features
with open(folder + r"\numerical_features.txt", "r") as file:
    numerical_cols = file.read().splitlines()

# Read categorical features
with open(folder + r"\categorical_features.txt", "r") as file:
    categorical_cols = file.read().splitlines()

print("Numerical Features:", len(numerical_cols))
print("Categorical Features:", len(categorical_cols))
```

##### **Section 4 — Remove Duplicates and Split Data**

```python
# Remove duplicate rows
train_df = train_df.drop_duplicates()

print("Dataset Shape:", train_df.shape)

# Random 80/20 split
train_data, test_data = train_test_split(
    train_df,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

print("Training Shape:", train_data.shape)
print("Testing Shape:", test_data.shape)
```

##### **Section 5 — Handle Structural Missing Values**

```python
structural_cols = [
    "Alley",
    "MasVnrType",
    "BsmtQual",
    "BsmtCond",
    "BsmtExposure",
    "BsmtFinType1",
    "BsmtFinType2",
    "FireplaceQu",
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "GarageCond",
    "PoolQC",
    "Fence",
    "MiscFeature"
]

for col in structural_cols:
    train_data[col] = train_data[col].fillna("None")
    test_data[col] = test_data[col].fillna("None")

print("Structural missing values handled.")
```

##### **Section 6 — Handle Remaining Missing Values**

```python
print("Categorical Missing Values:\n")

print("Training Dataset:")
print(train_data[categorical_cols].isnull().sum())

print("\nTesting Dataset:")
print(test_data[categorical_cols].isnull().sum())

print("\nTotal Missing Categorical Values:")
print("Training:", train_data[categorical_cols].isnull().sum().sum())
print("Testing:", test_data[categorical_cols].isnull().sum().sum())
```

```python
# Remove rows with missing categorical values only

train_data = train_data.dropna(subset=categorical_cols).copy()
test_data = test_data.dropna(subset=categorical_cols).copy()

print("Training Dataset Shape:", train_data.shape)
print("Testing Dataset Shape:", test_data.shape)

print("\nRemaining Categorical Missing Values:")
print("Training:", train_data[categorical_cols].isnull().sum().sum())
print("Testing:", test_data[categorical_cols].isnull().sum().sum())

print("\nMissing categorical rows removed successfully.")
```

##### **Section 7 — Ordinal and Binary Encoding**

```python
# LotShape
lot_shape_map = {
    "IR3": 1,
    "IR2": 2,
    "IR1": 3,
    "Reg": 4
}

# LandSlope
land_slope_map = {
    "Gtl": 1,
    "Mod": 2,
    "Sev": 3
}

# Quality mapping
quality_map = {
    "Po": 1,
    "Fa": 2,
    "TA": 3,
    "Gd": 4,
    "Ex": 5
}

# Basement quality
basement_quality_map = {
    "None": 0,
    "Po": 1,
    "Fa": 2,
    "TA": 3,
    "Gd": 4,
    "Ex": 5
}

# Basement exposure
basement_exposure_map = {
    "None": 0,
    "No": 1,
    "Mn": 2,
    "Av": 3,
    "Gd": 4
}

# Basement finish
basement_finish_map = {
    "None": 0,
    "Unf": 1,
    "LwQ": 2,
    "Rec": 3,
    "BLQ": 4,
    "ALQ": 5,
    "GLQ": 6
}

# Functional
functional_map = {
    "Sal": 1,
    "Sev": 2,
    "Maj2": 3,
    "Maj1": 4,
    "Mod": 5,
    "Min2": 6,
    "Min1": 7,
    "Typ": 8
}

# Fireplace quality
fireplace_map = {
    "None": 0,
    "Po": 1,
    "Fa": 2,
    "TA": 3,
    "Gd": 4,
    "Ex": 5
}

# Garage finish
garage_finish_map = {
    "None": 0,
    "Unf": 1,
    "RFn": 2,
    "Fin": 3
}

# Garage quality
garage_quality_map = {
    "None": 0,
    "Po": 1,
    "Fa": 2,
    "TA": 3,
    "Gd": 4,
    "Ex": 5
}

# Paved driveway
paved_drive_map = {
    "N": 0,
    "P": 1,
    "Y": 2
}

# Pool quality
pool_quality_map = {
    "None": 0,
    "Fa": 1,
    "TA": 2,
    "Gd": 3,
    "Ex": 4
}

# Binary features
central_air_map = {
    "N": 0,
    "Y": 1
}

street_map = {
    "Grvl": 0,
    "Pave": 1
}

# All mappings
ordinal_mappings = {
    "LotShape": lot_shape_map,
    "LandSlope": land_slope_map,
    "ExterQual": quality_map,
    "ExterCond": quality_map,
    "BsmtQual": basement_quality_map,
    "BsmtCond": basement_quality_map,
    "BsmtExposure": basement_exposure_map,
    "BsmtFinType1": basement_finish_map,
    "BsmtFinType2": basement_finish_map,
    "HeatingQC": quality_map,
    "KitchenQual": quality_map,
    "Functional": functional_map,
    "FireplaceQu": fireplace_map,
    "GarageFinish": garage_finish_map,
    "GarageQual": garage_quality_map,
    "GarageCond": garage_quality_map,
    "PavedDrive": paved_drive_map,
    "PoolQC": pool_quality_map,
    "CentralAir": central_air_map,
    "Street": street_map
}

# Apply mappings to both datasets
for col, mapping in ordinal_mappings.items():
    train_data[col] = train_data[col].map(mapping)
    test_data[col] = test_data[col].map(mapping)

print("Ordinal and binary encoding completed.")
```

##### **Section 8 — One-Hot Encoding**

```python
# Identify nominal categorical columns
nominal_cols = train_data.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Nominal Features:", nominal_cols)

# Apply One-Hot Encoding
train_data = pd.get_dummies(
    train_data,
    columns=nominal_cols,
    dtype=int
)

test_data = pd.get_dummies(
    test_data,
    columns=nominal_cols,
    dtype=int
)

# Make sure both datasets have identical columns
test_data = test_data.reindex(
    columns=train_data.columns,
    fill_value=0
)

print("One-Hot Encoding completed.")
```

##### **Section 9 — Handling Missing Numerical Values**

```python
# Identify columns with missing values in both datasets

train_missing = train_data.isnull().sum()
test_missing = test_data.isnull().sum()

missing_df = pd.DataFrame({
    "Data Type": train_data.dtypes,
    "Training Missing": train_missing,
    "Testing Missing": test_missing
})

# Keep only columns with missing values
missing_df = missing_df[
    (missing_df["Training Missing"] > 0) |
    (missing_df["Testing Missing"] > 0)
]

display(missing_df.sort_values(
    by="Training Missing",
    ascending=False
))
```

```python
# 1. Remove rows with missing GarageYrBlt
train_data = train_data.dropna(subset=["GarageYrBlt"]).copy()
test_data = test_data.dropna(subset=["GarageYrBlt"]).copy()


# 2. Median imputation for LotFrontage and MasVnrArea

for col in ["LotFrontage", "MasVnrArea"]:

    median_value = train_data[col].median()

    train_data[col] = train_data[col].fillna(median_value)
    test_data[col] = test_data[col].fillna(median_value)


# 3. Check remaining missing values

print("Remaining Missing Values:")

print(pd.DataFrame({
    "Training Missing": train_data.isnull().sum(),
    "Testing Missing": test_data.isnull().sum()
}).query("`Training Missing` > 0 or `Testing Missing` > 0"))
```

##### **Section 10 — Verify Preprocessing**

```python
print("Training Dataset Shape:", train_data.shape)
print("Testing Dataset Shape:", test_data.shape)

print("\nTraining Missing Values:",
      train_data.isnull().sum().sum())

print("Testing Missing Values:",
      test_data.isnull().sum().sum())

print("\nRemaining Categorical Columns:",
      len(train_data.select_dtypes(include=["object", "category"]).columns))

print("Feature Columns Match:",
      train_data.columns.equals(test_data.columns))
```

##### **Section 11 — Save Final Datasets**

```python
train_data.to_csv(
    folder + r"\clean_train.csv",
    index=False
)

test_data.to_csv(
    folder + r"\clean_test.csv",
    index=False
)

print("Clean training and testing datasets saved successfully!")
```
